7. Sum of weights

```python
cars_Asia = df[df['origin'] == 'Asia']

# cars_Asia
```

```python
vm = cars_Asia[['vehicle_weight','model_year']].iloc[:7]

# vm
```

```python
import numpy as np

X = vm.to_numpy()

# type(X)
```

```python
XTX = X.T @ X

XTX_inv = np.linalg.inv(XTX)

print(f"{XTX}XTX")
print()
print(f"{XTX_inv}XTX_inv")
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
c = np.linalg.inv(X.T @ X) @ X.T
```

```python
w = c @ y
```
